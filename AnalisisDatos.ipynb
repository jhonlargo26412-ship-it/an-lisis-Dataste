```python
# Reemplazar los valores faltantes por el promedio
df_hotel['Precio_Noche'] = df_hotel['Precio_Noche'].fillna(
    df_hotel['Precio_Noche'].mean()
)

# Crear la columna Costo_Hospedaje
df_hotel['Costo_Hospedaje'] = (
    df_hotel['Precio_Noche'] * df_hotel['Noches_Estadia']
)

# Mostrar los resultados
df_hotel.head()
```

```python
# Definir una función para calcular el costo del hospedaje
def calcular_costo(fila):
    return fila['Precio_Noche'] * fila['Noches_Estadia']

# Aplicar la función y crear una nueva columna
df_hotel['Costo_Hospedaje'] = df_hotel.apply(
    calcular_costo, axis=1
)

# Mostrar los resultados
df_hotel.head()
```

```python
# Renombrar las columnas
df_hotel = df_hotel.rename(columns={
    'Tipo_Habitacion': 'Habitacion',
    'Estado_Reserva': 'Estado'
})

# Reordenar las columnas
df_hotel = df_hotel[
    [
        'ID_Reserva',
        'Destino',
        'Habitacion',
        'Tipo_Huesped',
        'Precio_Noche',
        'Noches_Estadia',
        'Gasto_Adicional',
        'Estado',
        'Costo_Hospedaje'
    ]
]

# Mostrar la tabla reorganizada
df_hotel.head()
```

```python
# Agrupar por destino turístico
resumen_destinos = df_hotel.groupby('Destino').agg(
    Precio_Promedio=('Precio_Noche', 'mean'),
    Total_Costo_Hospedaje=('Costo_Hospedaje', 'sum')
)

# Mostrar los resultados
print(resumen_destinos)
```

```python
# Definir una función para calcular el rango
def calcular_rango(serie):
    return serie.max() - serie.min()

# Agrupar por tipo de habitación y calcular las métricas
dispersion = df_hotel.groupby('Habitacion').agg(
    Rango_Precio=('Precio_Noche', calcular_rango),
    Desviacion_Precio=('Precio_Noche', 'std'),
    Rango_Noches=('Noches_Estadia', calcular_rango),
    Desviacion_Noches=('Noches_Estadia', 'std')
)

# Mostrar los resultados
print(dispersion)
```

```python
# Agrupar por destino y tipo de habitación
resumen_multivariable = df_hotel.groupby(
    ['Destino', 'Habitacion']
).agg(
    Promedio_Noches=('Noches_Estadia', 'mean'),
    Total_Costo_Hospedaje=('Costo_Hospedaje', 'sum')
)

# Mostrar los resultados
print(resumen_multivariable)
```
